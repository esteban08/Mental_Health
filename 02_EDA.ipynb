# Etapas 1-2: Entendimiento del Negocio y los Datos (EDA)

En este cuaderno exploramos los datos limpios para extraer insights y patrones relacionados al riesgo de salud mental.

**Pasos a seguir:**
1. Carga de `cleaned_data.csv`.
2. Estadísticas descriptivas.
3. Histogramas y boxplots para numéricas.
4. Distribución de la variable objetivo.
5. Heatmap de correlaciones.
6. Análisis cruzado y pairplot de variables top.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid", palette="muted")
```

### 1. Carga de Datos y Estadísticas

```python
try:
    df = pd.read_csv('cleaned_data.csv')
    print("Datos limpios cargados.")
except FileNotFoundError:
    print("Error: No se encontró 'cleaned_data.csv'. Ejecuta 01_ETL.ipynb primero.")
    raise

display(df.describe())
```

### 2. Distribución de Variables Numéricas (Histogramas y Boxplots)

```python
num_cols = df.select_dtypes(include=['float64', 'int64']).columns

fig, axes = plt.subplots(len(num_cols), 2, figsize=(12, 4 * len(num_cols)))

for i, col in enumerate(num_cols):
    sns.histplot(df[col], kde=True, ax=axes[i, 0], color='#4caf8a')
    axes[i, 0].set_title(f'Histograma de {col}')
    
    sns.boxplot(x=df[col], ax=axes[i, 1], color='#9b8ec4')
    axes[i, 1].set_title(f'Boxplot de {col}')

plt.tight_layout()
plt.show()
```

### 3. Distribución de la Variable Objetivo `RISK_LEVEL`

```python
plt.figure(figsize=(8, 5))
ax = sns.countplot(data=df, x='RISK_LEVEL', order=['Bajo', 'Medio', 'Alto'], palette='viridis')
plt.title('Distribución de Clases: RISK_LEVEL')
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()), ha='center', va='bottom')
plt.show()
```

### 4. Correlaciones (Heatmap)

```python
plt.figure(figsize=(10, 8))
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Matriz de Correlación de Variables Numéricas')
plt.show()
```

### 5. Análisis Cruzado y Pairplot
Visualización de relación entre las principales variables y el `RISK_LEVEL`.

```python
# Top variables asumiendo que anxiety, depression, etc. influyen más.
top_vars = ['anxiety', 'depression', 'stress', 'panic_attacks', 'RISK_LEVEL']
sns.pairplot(df[top_vars], hue='RISK_LEVEL', palette='Set2', corner=True)
plt.suptitle('Pairplot de Variables Top vs RISK_LEVEL', y=1.02)
plt.show()
```
